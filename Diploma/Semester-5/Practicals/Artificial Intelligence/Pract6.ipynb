```python
# Importing the calendar module
import calendar

# Function to display the calendar for a given year
def display_calendar(year):
    try:
        # Ensure the year is a positive integer
        if year < 1:
            raise ValueError("Year must be a positive integer.")

        # Create a TextCalendar instance
        cal = calendar.TextCalendar(calendar.SUNDAY)

        # Print the full calendar for the given year
        print(f"\nFull calendar for the year {year}:\n")
        for month in range(1, 13):
            print(cal.formatmonth(year, month))
    except ValueError as e:
        print(f"Error: {e}")

# Function to get valid year input
def get_year():
    while True:
        try:
            year = int(input("Enter the year (e.g., 2024): "))
            return year
        except ValueError:
            print("Invalid input. Please enter a valid integer for the year.")

# Main logic to execute the program
year = get_year()
display_calendar(year)

```

```python
import calendar
from datetime import datetime
from IPython.display import display, HTML

# Function to display the calendar for a given year in HTML format
def display_calendar(year):
    try:
        # Ensure the year is a positive integer
        if year < 1:
            raise ValueError("Year must be a positive integer.")

        # Get the current date to highlight
        today = datetime.today()
        current_day = today.day if today.year == year else None
        current_month = today.month if today.year == year else None

        # Create HTML calendar for each month
        full_calendar_html = ""
        for month in range(1, 13):
            # Get the month's calendar as HTML and highlight the current day
            full_calendar_html += f"<h3>{calendar.month_name[month]} {year}</h3>"
            full_calendar_html += get_html_calendar(year, month, current_day if month == current_month else None)
            full_calendar_html += "<br>"

        # Display the final HTML calendar in the notebook
        display(HTML(full_calendar_html))

    except ValueError as e:
        print(f"Error: {e}")

# Function to get valid year input
def get_year():
    while True:
        try:
            year = int(input("Enter the year (e.g., 2024): "))
            return year
        except ValueError:
            print("Invalid input. Please enter a valid integer for the year.")

# Function to generate HTML calendar and highlight the current day
def get_html_calendar(year, month, current_day):
    cal = calendar.HTMLCalendar(calendar.SUNDAY)
    month_html = cal.formatmonth(year, month)

    # Highlight the current day by wrapping it in a span tag with style
    if current_day:
        highlight = f'<span style="color:red; font-weight:bold;">{current_day}</span>'
        month_html = month_html.replace(f'>{current_day}<', f'>{highlight}<')

    return month_html

# Main logic to execute the program
year = get_year()
display_calendar(year)
```
