```python
import datetime

def chatbot_response(user_input):
    user_input = user_input.lower()

    # Greeting
    if "hello" in user_input or "hi" in user_input:
        return "Hello! How can I assist you today?"

    # Asking about well-being
    elif "how are you" in user_input:
        return "I'm a chatbot, so I don't have feelings, but thanks for asking! How can I assist you?"

    # Time-related query
    elif "time" in user_input:
        current_time = datetime.datetime.now().strftime("%H:%M:%S")
        return f"The current time is {current_time}."

    # Date-related query
    elif "date" in user_input:
        current_date = datetime.datetime.now().strftime("%Y-%m-%d")
        return f"Today's date is {current_date}."

    # Weather-related query
    elif "weather" in user_input:
        return "I can't check the weather right now, but you can check your favorite weather app for the latest update."

    # Asking about the chatbot's name
    elif "your name" in user_input:
        return "I'm just a simple chatbot. You can call me your assistant!"

    # Asking about capabilities
    elif "can you do" in user_input:
        return "I can chat with you, provide basic information, and assist with simple tasks. What would you like to know?"

    # Farewell
    elif "bye" in user_input or "goodbye" in user_input:
        return "Goodbye! Have a great day!"

    # Default response for unknown queries
    else:
        return "I'm sorry, I don't understand that. Could you please rephrase your question?"

if __name__ == "__main__":
    print("Chatbot: Hello! How can I assist you today?")
    while True:
        user_input = input("You: ")
        if user_input.lower() == "exit":
            print("Chatbot: Goodbye! Have a great day!")
            break
        response = chatbot_response(user_input)
        print("Chatbot:", response)
```

```python
import datetime

def chatbot_response(user_input):
    try:
        # Convert user input to lowercase for case-insensitive matching
        user_input = user_input.lower().strip()

        if not user_input:
            return "It seems you didn't enter anything. Can you please type your question?"

        # Greeting
        if "hello" in user_input or "hi" in user_input:
            return "Hello! How can I assist you today?"

        # Asking about well-being
        elif "how are you" in user_input:
            return "I'm a chatbot, so I don't have feelings, but thanks for asking! How can I assist you?"

        # Time-related query
        elif "time" in user_input:
            current_time = datetime.datetime.now().strftime("%H:%M:%S")
            return f"The current time is {current_time}."

        # Date-related query
        elif "date" in user_input:
            current_date = datetime.datetime.now().strftime("%Y-%m-%d")
            return f"Today's date is {current_date}."

        # Weather-related query
        elif "weather" in user_input:
            return "I can't check the weather right now, but you can check your favorite weather app for the latest update."

        # Asking about the chatbot's name
        elif "your name" in user_input:
            return "I'm just a simple chatbot. You can call me your assistant!"

        # Asking about capabilities
        elif "can you do" in user_input:
            return "I can chat with you, provide basic information, and assist with simple tasks. What would you like to know?"

        # Farewell
        elif "bye" in user_input or "goodbye" in user_input:
            return "Goodbye! Have a great day!"

        # Default response for unknown queries
        else:
            return "I'm sorry, I don't understand that. Could you please rephrase your question?"

    except Exception as e:
        # Catch any unexpected error and return a friendly message
        return f"Oops! Something went wrong. Error: {str(e)}"

if __name__ == "__main__":
    print("Chatbot: Hello! How can I assist you today?")
    while True:
        try:
            user_input = input("You: ")
            if user_input.lower() == "exit":
                print("Chatbot: Goodbye! Have a great day!")
                break
            response = chatbot_response(user_input)
            print("Chatbot:", response)
        except KeyboardInterrupt:
            print("\nChatbot: Goodbye! Have a great day!")
            break
        except Exception as e:
            print(f"Chatbot: Sorry, something went wrong. {str(e)}")
```

```python
import datetime

# Store context (user information)
user_context = {
    "name": None,  # Store user's name
    "last_topic": None  # Store last topic discussed
}

def chatbot_response(user_input):
    global user_context
    user_input = user_input.lower().strip()

    # If the user introduces themselves, store their name
    if "my name is" in user_input:
        user_name = user_input.split("my name is")[-1].strip().capitalize()
        user_context["name"] = user_name
        return f"Nice to meet you, {user_name}!"

    # If the chatbot knows the user's name, use it in the response
    if user_context["name"]:
        greeting = f"Hello, {user_context['name']}! "
    else:
        greeting = "Hello! "

    # Greet the user
    if "hello" in user_input or "hi" in user_input:
        return greeting + "How can I assist you today?"

    # Asking about well-being
    elif "how are you" in user_input:
        return "I'm just a chatbot, but thanks for asking! How can I assist you?"

    # Time-related query
    elif "time" in user_input:
        current_time = datetime.datetime.now().strftime("%H:%M:%S")
        return f"The current time is {current_time}."

    # Date-related query
    elif "date" in user_input:
        current_date = datetime.datetime.now().strftime("%Y-%m-%d")
        return f"Today's date is {current_date}."

    # Weather-related query (simple placeholder)
    elif "weather" in user_input:
        user_context["last_topic"] = "weather"
        return "I can't check the weather right now, but you can check your favorite weather app."

    # Follow-up question about the last topic (e.g., weather)
    elif "more" in user_input and user_context["last_topic"] == "weather":
        return "I still can't check the weather, but it seems you're interested in it!"

    # Asking about the chatbot's name
    elif "your name" in user_input:
        return "I'm just a simple chatbot, but you can call me your assistant!"

    # Asking about the chatbot's capabilities
    elif "can you do" in user_input:
        user_context["last_topic"] = "capabilities"
        return "I can chat, provide time and date, and assist with basic queries. What would you like to know?"

    # Farewell
    elif "bye" in user_input or "goodbye" in user_input:
        return f"Goodbye, {user_context['name']}! Have a great day!" if user_context["name"] else "Goodbye! Have a great day!"

    # Default response for unknown queries
    else:
        return "I'm sorry, I don't understand that. Could you please rephrase your question?"

if __name__ == "__main__":
    print("Chatbot: Hello! How can I assist you today?")
    while True:
        try:
            user_input = input("You: ")
            if user_input.lower() == "exit":
                print("Chatbot: Goodbye! Have a great day!")
                break
            response = chatbot_response(user_input)
            print("Chatbot:", response)
        except KeyboardInterrupt:
            print("\nChatbot: Goodbye! Have a great day!")
            break
        except Exception as e:
            print(f"Chatbot: Sorry, something went wrong. {str(e)}")
```
